```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Bidirectional, SimpleRNN, LSTM, GRU, Dense, Input
```

```python
# load the IMDB dataset

num_words = 10000    # Numbers of words to keep
(X_train, y_train), (X_test, y_train) = imdb.load_data(num_words=num_words)
```

```python
# Pad sequences

maxlen = 100     # maximum sequence length

X_train = pad_sequences(X_train, maxlen=maxlen, padding='post', truncating='post')
X_test = pad_sequences(X_test, maxlen=maxlen, padding='post', truncating='post')
```

```python
# Building the RNN model :-

embedding_dim = 32        # Dimension of embedding layer

model = Sequential()

# Adding thw Input layer
model.add(Input(shape=(maxlen,)))

model.add(Embedding(input_dim=num_words, output_dim=embedding_dim))
model.add(SimpleRNN(5))                          # add 5 RNN units
model.add(Dense(1, activation='sigmoid'))        # Binary Classification (positive / negative)

# compile the model
model.compile(optimizer='adam', loss='bianry_crossentropy', metrics=['accuracy'])

# Display model architecture
model.summary()
```

```python
# Similarly on the same basis, we can also create our "Bi-directional model" : -
```

```python
# Building the Bidirectional RNN model :-

model = Sequential()

# Adding thw Input layer
model.add(Input(shape=(maxlen,)))

model.add(Embedding(input_dim=num_words, output_dim=embedding_dim))
model.add(Bidirectional(SimpleRNN(5)))                          # Add 5 RNN units
model.add(Dense(1, activation='sigmoid'))        # Binary Classification (positive / negative)

# compile the model
model.compile(optimizer='adam', loss='bianry_crossentropy', metrics=['accuracy'])

# Display model architecture
model.summary()
```

```python
# In this, our weight and biases both becomes doubled as we are using Bi-directional model.
# Also we are using 2 RNNS, instead of 1 RNN.
```

```python
# Similarly, we can also apply this concept(i.e., Bi-directional) on the LSTM and GRU model 
```

```python
# Building the Bidirectional LSTM model :- 

embedding_dim = 32        # Dimension of embedding layer

model = Sequential()

# Adding thw Input layer
model.add(Input(shape=(maxlen,)))

model.add(Embedding(input_dim=num_words, output_dim=embedding_dim))
model.add(Bidirectional(LSTM(5)))                          # add 5 RNN units
model.add(Dense(1, activation='sigmoid'))        # Binary Classification (positive / negative)

# compile the model
model.compile(optimizer='adam', loss='bianry_crossentropy', metrics=['accuracy'])

# Display model architecture
model.summary()
```

```python
# Building the Bidirectional GRU model :- 

model = Sequential()

# Adding thw Input layer
model.add(Input(shape=(maxlen,)))

model.add(Embedding(input_dim=num_words, output_dim=embedding_dim))
model.add(Bidirectional(GRU(5)))                          # add 5 RNN units
model.add(Dense(1, activation='sigmoid'))        # Binary Classification (positive / negative)

# compile the model
model.compile(optimizer='adam', loss='bianry_crossentropy', metrics=['accuracy'])

# Display model architecture
model.summary()
```

```python
# So here we can clearly see that here weights are 'increased' inn both model
# b/c in LSTM and GRU, there are more weights and biases present compared to RNN.
```
